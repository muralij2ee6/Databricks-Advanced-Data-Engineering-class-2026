# Databricks Data Privacy
#### Duration: 5 minutes

## Overview

In a typical lakehouse, personal data spreads from source tables into downstream copies and aggregates. Obligations such as restricted access and the right to erasure hold only when every copy is found, protected by purpose and risk and accounted for with evidence. Databricks Data Privacy treats privacy as a set of technical decisions and controls built on Unity Catalog and Delta Lake. It separates privacy from security, governance and compliance, then organizes the work around an identify, protect, manage and verify lifecycle and a decision model applied to every technique.

The course follows one retailer's data estate through that lifecycle. Data Classification and governed tags identify sensitive columns and record how each should be treated. Tag-driven attribute-based access control (ABAC) policies mask those columns across a schema, with manual column masks and row filters for the cases a central policy can't cover. De-identification goes further with salted hashing, tokenization through a separately governed vault and column encryption, with the salt and encryption key held in Databricks Secrets. For the right to erasure, the course tracks a request in a control plane, finds the subject's data with classification and lineage, deletes and propagates the change with `DELETE` and `MERGE`, then purges the underlying files with `REORG TABLE ... APPLY (PURGE)` and retention-aware `VACUUM`. Lineage, audit logs, query history and Delta table history then combine into an evidence trail an auditor can check.

The course is technical and does not provide legal advice.

## Terminal Objectives

By the end of this course, you will be able to:

1. **Explain** how privacy differs from security, governance and compliance and how the identify, protect, manage and verify lifecycle organizes privacy work on Databricks.
2. **Choose** a privacy control for a given purpose and risk using a structured privacy decision model.
3. **Use** Data Classification and governed tags to find sensitive columns and record how each should be treated.
4. **Configure** tag-driven ABAC policies, manual column masks and row filters to restrict access to sensitive columns.
5. **Apply** de-identification techniques, including salted hashing, tokenization with a vault and column encryption, with salts and keys stored in Databricks Secrets.
6. **Execute** a right-to-erasure request from discovery with classification and lineage through deletion, downstream propagation and physical file purge.
7. **Construct** an evidence trail from classification, lineage, audit logs, query history and Delta table history that shows whether a privacy control worked.

## A. Prerequisites

Before starting this course, learners should be comfortable with the following:

- Basic familiarity with the Databricks workspace, including navigating it, opening notebooks and running notebook cells.
- Familiarity with Unity Catalog concepts, including catalogs, schemas, tables and granting privileges.
- Intermediate SQL, including joins, aggregations and `MERGE` statements.
- Basic familiarity with Delta Lake tables, including table history and time travel.
- Familiarity with layered data pipelines such as the medallion (bronze, silver, gold) architecture.
