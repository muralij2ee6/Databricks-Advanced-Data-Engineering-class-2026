# Privacy, Security, Governance and Compliance
#### Duration: 15 minutes

## Overview

In this lecture we'll frame what privacy means on Databricks before you start building controls. 

People use <b>privacy</b>, <b>security</b>, <b>governance</b> and <b>compliance</b> as if they were the same thing, so this lecture separates them first, then shows where they overlap.

From there you meet the lifecycle the whole course reuses, identify, protect, manage and verify and the privacy decision model, a short set of questions you run on every technique. 

We'll turn vague privacy obligations into concrete technical questions you can answer on Databricks and sets the compliance framing. 

Keep in mind that this course is <b>non-legal</b> and the live Trust and Compliance Center is the source of truth for certifications.

> Stance: privacy is a set of decisions and controls you implement and verify on Databricks, not a compliance certificate.

## Learning Objectives

By the end of this lecture, you'll be able to:

1. **Distinguish** privacy from security, governance and compliance and **explain** where they overlap.
2. **Explain** the identify, protect, manage and verify lifecycle and the role each phase plays.
3. **Translate** privacy obligations into technical questions you can answer on Databricks.
4. **Describe** how the certifications Databricks holds differ from the regulations it helps you meet and **locate** the live Trust and Compliance Center.
5. **Reuse** the running scenario and the six-question privacy decision model that every later demo and lab returns to.

### A1. Privacy draws on the other three

Privacy isn't a separate silo. It uses security to protect personal data, governance to know where that data lives and compliance to know which obligations apply. 
Privacy sits where the three meet and it adds its own focus: <b>personal data and individual rights</b>.

##### EXPAND FOR ADDITIONAL DETAILS

<details>
<summary style="font-size:14pt; font-weight:600; color:#1B5162; cursor:pointer;">Why the distinction matters</summary>

<div style="font-size:14pt; line-height:1.6; color:#0b2026; padding:12px 0;">

- **Security is necessary but not sufficient for privacy.** A perfectly secured table can still expose personal data to people inside the company who shouldn't see it and it can still hold data you were supposed to delete. Privacy adds purpose and lifecycle on top of access control.
- **Governance is the substrate.** You can't protect or delete personal data you haven't cataloged and can't locate. Discovery and classification (Demo 03) is a governance capability the privacy work depends on.
- **Compliance and privacy are not the same.** A regulation says a person can request erasure. Privacy is the technical work that finds the data, deletes it everywhere and proves it.
- **The trap to avoid:** treating a compliance certificate as proof of privacy. A certificate describes the platform's posture, not whether your specific tables are handled correctly. That's your job and it's what you'll learn to do here.

</div>
</details>

##### EXPAND FOR ADDITIONAL DETAILS

<details>
<summary style="font-size:14pt; font-weight:600; color:#1B5162; cursor:pointer;">What each phase means</summary>

<div style="font-size:14pt; line-height:1.6; color:#0b2026; padding:12px 0;">

- **Identify:** find the personal data across the estate. Demo 03 uses Data Classification and governed tags.
- **Protect:** restrict or transform it by purpose. Demo 04 uses tag-driven ABAC and manual masks. Demo 06 uses de-identification and key management.
- **Manage:** correct, delete and purge personal data over its lifecycle. Demo 07 is the erasure workflow. (De-identification and key management sit under Protect, in Demo 06, because they transform the value to protect it.)
- **Verify:** produce evidence that the control worked. Demo 08 assembles it from system tables, lineage and history.
- **Why verify matters:** applying a control isn't the end of the job. A regulator, an auditor or a customer will ask you to show that a person's data was actually protected or deleted. Every demo and lab in this course ends by producing that evidence, so verify is a first-class phase rather than an afterthought.

</div>
</details>

##### EXPAND FOR ADDITIONAL DETAILS

<details>
<summary style="font-size:14pt; font-weight:600; color:#1B5162; cursor:pointer;">Why restate obligations this way</summary>

<div style="font-size:14pt; line-height:1.6; color:#0b2026; padding:12px 0;">

- A plain obligation like "respect the right to erasure" is hard to act on directly. The technical question "can we delete this subject everywhere their data spread and prove it" tells you exactly what to build and how to check it.
- The four questions line up with the four phases, so the lifecycle isn't abstract. Each phase exists to answer one of these questions on your own data.
- These four obligations are the plain-language starting point. Section E1 turns them into the six-question decision model every demo runs and each demo ends by answering the last one, proving what happened.

</div>
</details>
