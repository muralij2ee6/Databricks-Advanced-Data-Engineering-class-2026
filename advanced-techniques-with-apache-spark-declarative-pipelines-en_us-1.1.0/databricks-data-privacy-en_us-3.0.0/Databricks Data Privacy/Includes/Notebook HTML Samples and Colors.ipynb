# DE-PRO-2 Notebook HTML Samples and Colors

Reference cells for authors. The approved palette and the callout/admonition
patterns below match the shared curriculum design system. Compose slides from
the Card-Template-Catalog blocks. Never invent colors, fonts, sizes or spacing.

Rules:
- Always use `%md-sandbox` for HTML (regular `%md` strips it).
- 14pt is the absolute minimum font size.
- Accent rotation on cards: blue -> green -> coral -> amber -> red.
- No em-dashes or en-dashes. No AI icons or rocket ships.

## Expandable script pattern

Do not title it "Instructor Script". No AI icons.
